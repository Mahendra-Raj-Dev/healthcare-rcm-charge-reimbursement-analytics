```python
import sqlite3
import pandas as pd

with sqlite3.connect(r"D:\Data Analytics\healthcare-rcm-charge-reimbursement-analytics\healthcare_rcm.db") as conn:
    hospitals = pd.read_sql_query("SELECT * FROM hospitals", conn)
    drgs = pd.read_sql_query("SELECT * FROM drgs", conn)
    fact_discharges = pd.read_sql_query("SELECT * FROM fact_discharges", conn)
```

```python
hospitals.to_csv(r"../data/hospitals.csv", index=False)
drgs.to_csv(r"../data/drgs.csv", index=False)
fact_discharges.to_csv(r"../data/fact_discharges.csv", index=False)
```
